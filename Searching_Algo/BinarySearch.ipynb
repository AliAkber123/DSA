# Binary Search — Interview Revision Notes

## 1. When to Use Binary Search
- When the array is **sorted**.
- When you need to find a target efficiently.
- Instead of checking every element, eliminate half of the search space each iteration.

## 2. Algorithm — Remember These 5 Steps

1. Initialize `left = 0`, `right = len(nums) - 1`.
2. Loop while `left <= right`.
3. Calculate `mid = (left + right) // 2`.
4. Compare `nums[mid]` with `target`:
   - Equal → Return `mid`.
   - Target greater → `left = mid + 1`.
   - Target smaller → `right = mid - 1`.
5. If not found → Return `-1`.



## 3. Important Things to Remember

| Concept | Remember |
|---|---|
| `left <= right` | Includes the final remaining element |
| `// 2` | Integer division; no `math.floor()` needed |
| `left = mid + 1` | Discard middle and everything to its left |
| `right = mid - 1` | Discard middle and everything to its right |
| `return -1` | Target not found |
| Sorted array | Required for standard Binary Search |



## 4. Complexity

- **Time: O(log n)** — eliminates half of the remaining elements each iteration.
- **Space: O(1)** — only uses `left`, `right`, and `mid`.


Q1: 
https://leetcode.com/problems/binary-search/submissions/2184194784/

```python

class Solution:
    def search(self, nums: list[int], target: int) -> int:
        left=0
        right=len(nums)-1

        while left <= right:
            mid = (left + right)//2
            if target==nums[mid]:
                return mid
            elif target > nums[mid]:
                left=mid+1
            else:
                right=mid-1
        return -1
```
