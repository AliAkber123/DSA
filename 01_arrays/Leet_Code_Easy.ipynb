Q1:
https://leetcode.com/problems/two-sum/description/

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        output=[]
        for i in range(len(nums)):
            for j in range(i+1,len(nums)):
                if nums[i]+nums[j] == target:
                    output.append(i)
                    output.append(j)
        return output 
```

Q2:
https://leetcode.com/problems/running-sum-of-1d-array/submissions/2182739140/

```python
class Solution:
    def runningSum(self, nums: list[int]) -> list[int]:
        sum=0
        output=[]
        for i in range(len(nums)):
            sum=sum+nums[i]
            output.append(sum)

        return output
```

Q3:
https://leetcode.com/problems/concatenation-of-array/

```python
class Solution:
    def getConcatenation(self, nums: list[int]) -> list[int]:
        ans=[]
        
        for i in nums:
            ans.append(i)
        for i in nums:
            ans.append(i)
        return ans
```

Q4: Shuffle the Array 

https://leetcode.com/problems/shuffle-the-array/

```python
class Solution:
    def shuffle(self, nums: List[int], n: int) -> List[int]:
        output=[]

        for i in range(n):
            j = i+n
            output.append(nums[i])
            output.append(nums[j])
        return output
        
```

Q5:
Concept: 2D arrays

https://leetcode.com/problems/richest-customer-wealth/

```python
class Solution:
    def maximumWealth(self, accounts: list[list[int]]) -> int:
        rich=[]
        
        for i in range(len(accounts)):
            wealth=0
            for j in range(len(accounts[i])):
                wealth = wealth + accounts[i][j]
            rich.append(wealth)
        
        
        rich_person=rich[0]
        for i in range(len(rich)):
            
            if rich[i] >= rich_person:
                rich_person=rich[i]
        return rich_person
```

Q6: Find Numbers with Even Number of Digits

https://leetcode.com/problems/find-numbers-with-even-number-of-digits/submissions/2183219878/

```python
class Solution:
    def findNumbers(self, nums: list[int]) -> int:
        count=0

        for i in range(len(nums)):
            if len(str(nums[i])) %2 == 0:
                count=count+1
        return count
        
```

Q7: Max Consecutive Ones

https://leetcode.com/problems/max-consecutive-ones/

```python
class Solution:
    def findMaxConsecutiveOnes(self, nums: list[int]) -> int:
        count = 0
        maxi=0
        for i in nums:
            if i == 1:
                count=count+1
                if count>maxi:
                    maxi=count                
            else:
                
                count=0
            
        return maxi

        
```
